Data Exploration

```python
import pandas as pd
import numpy as np
import os
```

```python
files = [
    "dailyCalories_merged.csv",
    "hourlyCalories_merged.csv",
    "minuteCaloriesNarrow_merged.csv",
    "minuteCaloriesWide_merged.csv"
]

print(files)
```

```python
summary = []

for file in files:
    df = pd.read_csv(file)

    # Print exploration info
    print(f"\n{file}")
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Missing values:", df.isnull().sum().sum())
    print("Column names:", df.columns.tolist())

    # Store summary info
    summary.append({
        "File": file,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Users": df["Id"].nunique() if "Id" in df.columns else "N/A"
    })
```

```python
for file in files:
    df = pd.read_csv(file)

    print(file)

    print("\nData Types:")
    print(df.dtypes)

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nSample Data:")
    print(df.head())
```

```python
for file in files:
    df = pd.read_csv(file)

    if "ActivityDay" in df.columns:
        df["ActivityDay"] = pd.to_datetime(
            df["ActivityDay"],
            format="%m/%d/%Y"
        )

    elif "ActivityHour" in df.columns:
        df["ActivityHour"] = pd.to_datetime(
            df["ActivityHour"],
            format="%m/%d/%Y %I:%M:%S %p"
        )

    elif "ActivityMinute" in df.columns:
        df["ActivityMinute"] = pd.to_datetime(
            df["ActivityMinute"],
            format="%m/%d/%Y %I:%M:%S %p"
        )

    print(f"\n{file}")
    print(df.dtypes)
```

```python
for file in files:
    df = pd.read_csv(file)

    date_col = [c for c in df.columns if "Activity" in c][0]

    dates = pd.to_datetime(df[date_col])

    print(file)
    print("Start:", dates.min())
    print("End:", dates.max())
    print()
```

Key Findings:

- No missing values across any of the four datasets
- No duplicate records identified
- All datasets contain a user ID, making them suitable for user-level analysis
- The datasets provide calorie data at different levels of granularity:
  . dailyCalories_merged.csv provides daily summaries.
  . hourlyCalories_merged.csv provides hourly trends.
  . minuteCaloriesNarrow_merged.csv provides detailed minute-by-minute records in a long format.
  . minuteCaloriesWide_merged.csv contains minute-level data in a wide format.

- The same metric (calories burned) is represented across different time scales, making the datasets suitable for drill-down analysis within a dashboard.

Data Cleaning Requirements:

- The main data cleaning requirement identified was converting the date/time fields from string (object) format to datetime format. This has now been successfully applied across all four datasets, enabling time-series analysis, filtering, aggregation, and dashboard visualisations.

Data Suitability:

- The datasets are well-structured and analysis-ready following datetime conversion.
- Consistent user IDs across all files enable integration and comparison between different levels of granularity.
- The data is suitable for both descriptive analysis (what happened) and trend analysis (how calorie expenditure changes over time).
- The presence of daily, hourly, and minute-level data supports dashboard drill-down functionality.

Dashboard Use Cases:

- Track average calories burned per day.
- Compare calorie expenditure between users.
- Identify peak calorie-burning hours throughout the day.
- Analyse daily and hourly calorie trends over time.
- Allow users to drill down from daily summaries into hourly and minute-level details.
- Create interactive time-based visualisations using the converted datetime fields.
- Support user segmentation based on activity and calorie expenditure patterns.
- Use dailyCalories for summary KPIs and trend reporting.
- Use hourlyCalories for behavioural analysis and peak activity visualisations.
- Use minuteCaloriesNarrow only for detailed drill-through pages due to dataset size.
- Avoid using minuteCaloriesWide directly in visualisations unless it is first transformed into a long format.
- Include date filters and user filters to allow interactive exploration of activity patterns.

Challenges:

- minuteCaloriesNarrow_merged.csv contains a very large number of records, which could impact dashboard performance if loaded directly.
- minuteCaloriesWide_merged.csv is not in an analysis-friendly format and may require restructuring if detailed minute-level analysis is needed.
- Multiple datasets contain the same information at different levels of granularity, so care will be needed to avoid double-counting or inconsistent measures within the dashboard.
- The datasets only capture calorie expenditure and do not provide contextual information about the activities responsible for the calories burned.

Conclusion:

- Overall, the four allocated datasets were found to be of good quality, with no missing values or duplicate records identified. The primary data preparation task involved converting date and time fields to datetime format. Following this transformation, the datasets were deemed suitable for dashboard development, providing multiple levels of detail that can support both high-level reporting and detailed activity analysis through drill-down functionality.

### 1. Streamlit vs Power BI

After looking into both options, my current recommendation would be **Streamlit** if we're aiming to build a software product and demonstrate development skills, whereas **Power BI** would be better if our main goal is to create a dashboard quickly with minimal coding.

Streamlit

- Python-based and integrates well with data analysis workflows.
- Allows much more flexibility and custom features.
- Good opportunity to demonstrate software development practices (requirements, version control, testing, deployment).
- Steeper learning curve than Power BI.

Estimated time for me to get up to speed

- Basic dashboard: ~4 hours
- Comfortable building interactive dashboards: ~1–2 weeks of project work

Power BI

- Faster to produce dashboards.
- Easier to learn initially.
- Strong visualisation and filtering functionality out of the box.
- Less flexible for custom application features.

Estimated time for me to get up to speed

- Basic dashboard creation: ~3 hours
- Comfortable with interactive reports: ~1 week

My view at the moment is that **Streamlit aligns better with the software development side of the project**, but happy to discuss as a team.
